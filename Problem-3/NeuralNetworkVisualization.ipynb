```python
import matplotlib.pyplot as plt

# Create figure
fig, ax = plt.subplots(figsize=(12, 8))

# -----------------------------
# Layer positions
# -----------------------------

input_x = 1
hidden1_x = 4
hidden2_x = 7
output_x = 10

input_y = [6,5,4,3,2,1]
hidden1_y = [7,6,5,4,3,2,1,0]
hidden2_y = [5.5,4,2.5,1]
output_y = [3.5]

input_labels = [
    "Transaction Amount",
    "Transaction Time",
    "Merchant Category",
    "Customer Age",
    "Account Balance",
    "Transactions Today"
]

# -----------------------------
# Draw Input Layer
# -----------------------------
for i, y in enumerate(input_y):
    ax.scatter(input_x, y, s=900, color='skyblue', edgecolors='black')
    ax.text(input_x-0.2, y,
            input_labels[i],
            ha='right',
            va='center',
            fontsize=10)

# -----------------------------
# Draw Hidden Layer 1
# -----------------------------
for y in hidden1_y:
    ax.scatter(hidden1_x, y, s=900, color='orange', edgecolors='black')

# -----------------------------
# Draw Hidden Layer 2
# -----------------------------
for y in hidden2_y:
    ax.scatter(hidden2_x, y, s=900, color='orange', edgecolors='black')

# -----------------------------
# Draw Output Layer
# -----------------------------
ax.scatter(output_x,
           output_y[0],
           s=1000,
           color='lightgreen',
           edgecolors='black')

ax.text(output_x+0.4,
        output_y[0],
        "Fraud\n(0/1)",
        fontsize=11)

# -----------------------------
# Draw Connections
# -----------------------------

for y1 in input_y:
    for y2 in hidden1_y:
        ax.plot([input_x, hidden1_x],
                [y1, y2],
                color="gray",
                linewidth=0.4)

for y1 in hidden1_y:
    for y2 in hidden2_y:
        ax.plot([hidden1_x, hidden2_x],
                [y1, y2],
                color="gray",
                linewidth=0.4)

for y in hidden2_y:
    ax.plot([hidden2_x, output_x],
            [y, output_y[0]],
            color="gray",
            linewidth=0.5)

# -----------------------------
# Layer Names
# -----------------------------

ax.text(input_x,7.8,
        "Input Layer",
        fontsize=12,
        fontweight='bold',
        ha='center')

ax.text(hidden1_x,7.8,
        "Hidden Layer 1",
        fontsize=12,
        fontweight='bold',
        ha='center')

ax.text(hidden2_x,7.8,
        "Hidden Layer 2",
        fontsize=12,
        fontweight='bold',
        ha='center')

ax.text(output_x,7.8,
        "Output Layer",
        fontsize=12,
        fontweight='bold',
        ha='center')

plt.title("Neural Network Architecture for Credit Card Fraud Detection",
          fontsize=16,
          fontweight='bold')

ax.set_xlim(0,11)
ax.set_ylim(-1,8)

ax.axis("off")

plt.show()
```
