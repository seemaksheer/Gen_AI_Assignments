```python
# Employee Productivity Prediction using Multivariate Linear Regression
# Import pandas to create and work with tables (DataFrames)
import pandas as pd
```

```python
# Import the Linear Regression algorithm
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
# Import evaluation metrics
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Create the dataset using a Python dictionary.
# Each key represents a column in the dataset.
data = {
    "Experience": [2, 5, 1, 8, 4, 10, 3, 6, 7, 2],
    "TrainingHours": [40, 60, 20, 80, 50, 90, 30, 70, 75, 25],
    "WorkingHours": [38, 42, 35, 45, 40, 48, 37, 44, 46, 36],
    "Projects": [3, 6, 2, 8, 5, 9, 4, 7, 7, 3],
    "Productivity": [62, 78, 55, 88, 72, 92, 65, 82, 85, 60]
}
```

```python
# Convert the dictionary into a DataFrame
df = pd.DataFrame(data)
```

```python
# Display the complete dataset
print("Employee Dataset")
print(df)
```

```python
# X contains the independent variables
X = df[["Experience", "TrainingHours", "WorkingHours", "Projects"]]
```

```python
# y contains the dependent variable that we want to predict
y = df["Productivity"]
```

```python
# Display the selected features
print("\nInput Features (X)")
print(X)
```

```python
# Display the target variable
print("\nTarget Variable (y)")
print(y)
```

```python
# Use 80% data for training & 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
print("\nTraining Records :", len(X_train))
print("Testing Records :", len(X_test))
```

```python
# Create the machine learning model
model = LinearRegression()

# Train the model using the training dataset
model.fit(X_train, y_train)

print("\nModel trained successfully.")
```

```python
# Create the machine learning model
model = LinearRegression()

# Train the model using the training dataset
model.fit(X_train, y_train)

print("\nModel trained successfully.")
```

```python
# Print the intercept (constant value)
print("\nIntercept:")
print(model.intercept_)

# Print the coefficient of every feature
print("\nRegression Coefficients")

for feature, coefficient in zip(X.columns, model.coef_):
    print(feature, "=", round(coefficient, 4))
```

```python
# Predict productivity scores
predicted = model.predict(X_test)

print("\nActual Productivity")
print(list(y_test))

print("\nPredicted Productivity")
print(predicted)
```

```python
# Calculate Mean Squared Error
mse = mean_squared_error(y_test, predicted)

# Calculate R2 Score
r2 = r2_score(y_test, predicted)

print("\nModel Evaluation")

print("Mean Squared Error =", round(mse, 2))
print("R2 Score =", round(r2, 2))
```

```python
# Create details of a new employee
new_employee = pd.DataFrame(
    [[5, 65, 42, 6]],
    columns=["Experience", "TrainingHours", "WorkingHours", "Projects"]
)

# Predict productivity
new_prediction = model.predict(new_employee)

print("\nPredicted Productivity for New Employee")
print(round(new_prediction[0], 2))
```
